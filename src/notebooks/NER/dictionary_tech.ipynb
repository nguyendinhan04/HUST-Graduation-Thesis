```python
import pandas as pd
```

```python
df = pd.read_csv('../../../data/job_descriptions.csv')
```

```python
df
```

```python
import ast
def parse_key_technology(key_technology_str):
    try:
        return ast.literal_eval(key_technology_str)
    except (ValueError, SyntaxError):
        return []
    

df['key_technologies'] = df['key_technologies'].apply(parse_key_technology)
```

```python
df
```

```python
key_tech_series = df['key_technologies'].explode().reset_index(drop=True)
key_tech_series = key_tech_series.dropna()
key_tech_df = key_tech_series.to_frame(name='key_technologies')
```

```python
key_tech_df
```

```python
unique_key_technologies_df = key_tech_df["key_technologies"].str.strip().str.lower().drop_duplicates()

```

```python
unique_key_technologies_df
```

```python
unique_key_technologies_df.to_csv('../../../data/ready_data/unique_key_technologies.csv', index=False)
```

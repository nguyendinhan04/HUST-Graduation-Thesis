# Parse detail salary

```python
salary_df = df["detail_salary"].copy()
salary_df = salary_df.drop_duplicates().reset_index(drop=True)
salary_df = salary_df.to_frame()
salary_df
```

```python
import re
def parse_salary(salary_str):
    if pd.isna(salary_str):
        return None, None,None

    salary_str = salary_str.replace(",", "").strip()
    if "Thỏa thuận" in salary_str:
        return None, None,None
    
    currency = None
    if "USD" in salary_str:
        currency = "USD"
    if "triệu" in salary_str: 
        currency = "trieu VND"
    
    numbers = [float(s) for s in re.findall(r'\d+(?:.\d+)?', salary_str)]

    if "Từ" in salary_str:
        min_salary = numbers[0]
        return min_salary, None,currency
    
    if "Tới" in salary_str:
        max_salary = numbers[0]
        return None, max_salary,currency
    
    if "-" in salary_str:
        min_salary = numbers[0]
        max_salary = numbers[1]
        return min_salary, max_salary,currency
    
    return None, None, None
```

```python
print(parse_salary("Tới 2,000 USD"))  # (500.0, None)
```

```python
salary_df['min_salary'], salary_df['max_salary'],salary_df['currency'] = zip(*salary_df['detail_salary'].apply(parse_salary))   
```

# Parse experience

```python
expericence_df = df["detail_experience"].copy()
expericence_df = expericence_df.drop_duplicates().reset_index(drop=True)
expericence_df = expericence_df.to_frame()
expericence_df
```

Có thể là những job dưới 1 năm kinh nghiệm là fresher
con những job không yêu cầu kinh nghiệm là internship

```python
def parse_experience(exp_str):
    if pd.isna(exp_str):
        return None

    exp_str = exp_str.replace("năm", "").strip()
    if "Không yêu cầu" in exp_str:
        return 0

    if "Dưới 1" in exp_str:
        return 0.5
    
    if "Trên 5" in exp_str:
        return 7
    
    numbers = [float(s) for s in re.findall(r'\d+(?:\.\d+)?', exp_str)]

    return numbers[0]
```

```python
expericence_df["min_experience"] = expericence_df['detail_experience'].apply(parse_experience)
expericence_df
```

# Parse Working time

```python
wt_df = df["working_times"].copy()
wt_df = wt_df.drop_duplicates().reset_index(drop=True)
wt_df = wt_df.to_frame()
wt_df
```

```python
import re
def parse_working_time(wt_str):
    if pd.isna(wt_str):
        return None

    wt_str = wt_str.strip().lower()
    if "thứ 2 - thứ 6" in wt_str:
        return 5
    if "thứ 2 - thứ 7" in wt_str:
        return 6
    if "thứ 2 - chủ nhật" in wt_str:
        return 7
    return "Other"
```

```python
wt_df["parsed_working_days"] = wt_df['working_times'].apply(parse_working_time)
wt_df
```

```python
df["parsed_working_days"] = df["working_times"].apply(parse_working_time)
df 
```

# Parse general info

```python
general_info_df = df[["general_info"]].copy()
general_info_df
```

```python
import ast
def parse_general_info(info_str):
    if pd.isna(info_str):
        return {}
    try:
        info_dict = ast.literal_eval(info_str)
        if isinstance(info_dict, dict):
            return info_dict
        else:
            return {}
    except (ValueError, SyntaxError):
        return {}
```

```python
general_info_df["parsed_general_info"] = general_info_df['general_info'].apply(parse_general_info)
general_info_df
```

```python
general_info_expand = general_info_df['parsed_general_info'].apply(pd.Series)
```

```python
general_info_df = pd.concat([general_info_df, general_info_expand], axis=1)
general_info_df
```

# Parse box categories

```python
box_categories_df = df[["box_categories"]].copy()
box_categories_df
```

```python
def parse_box_categories(box_str):
    if pd.isna(box_str):
        return {}
    try:
        parse_box = {}
        box_list = ast.literal_eval(box_str)
        if isinstance(box_list, list):
            for item in box_list:
                if isinstance(item, dict) and 'box_title' in item and 'categories' in item:
                    parse_box.update({item['box_title'] : item['categories']})
            return parse_box
        else:
            return {}
    except (ValueError, SyntaxError):
        return {}
```

```python
box_categories_df["box_categories_parsed"] = box_categories_df['box_categories'].apply(parse_box_categories)
box_categories_df
```

```python
box_categories_expand_df= box_categories_df['box_categories_parsed'].apply(pd.Series)
box_categories_expand_df
```

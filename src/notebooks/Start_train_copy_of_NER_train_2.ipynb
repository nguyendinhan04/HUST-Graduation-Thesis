```python
label_names = ["O", "B-TECH_SKILL", "I-TECH_SKILL", "B-KNOW", "I-KNOW"]
id2label = {i: label for i, label in enumerate(label_names)}
label2id = {label: i for i, label in enumerate(label_names)}

assert all(label == label.strip() for label in label_names), label_names
print(label_names)
```

```python
import ast
import re
import unicodedata
from collections import Counter

import pandas as pd

TOKEN_PATTERN = re.compile(r"\w+|[^\w\s]", flags=re.UNICODE)


def normalize_text(value):
    """Normalize text before token matching so labels are less noisy."""
    value = unicodedata.normalize("NFC", str(value)).strip().lower()
    value = re.sub(r"\s+", " ", value)
    return value


def tokenize_text(value):
    return TOKEN_PATTERN.findall(unicodedata.normalize("NFC", str(value)))


def ensure_list(data):
    if isinstance(data, list):
        return data
    if data is None or pd.isna(data):
        return []
    if isinstance(data, str):
        try:
            parsed = ast.literal_eval(data)
            return parsed if isinstance(parsed, list) else []
        except (SyntaxError, ValueError):
            return []
    return []


def convert_to_bio_format(text, tech_skill, knowledge):
    tokens = tokenize_text(text)
    normalized_tokens = [normalize_text(token) for token in tokens]
    labels = [label2id["O"]] * len(tokens)
    conflicts = []

    keyword_specs = []
    for entity_type, keywords in [
        ("TECH_SKILL", ensure_list(tech_skill)),
        ("KNOW", ensure_list(knowledge)),
    ]:
        for keyword in keywords:
            keyword_tokens = tokenize_text(keyword)
            normalized_keyword_tokens = [normalize_text(token) for token in keyword_tokens]
            normalized_keyword_tokens = [token for token in normalized_keyword_tokens if token]
            if normalized_keyword_tokens:
                keyword_specs.append((
                    len(normalized_keyword_tokens),
                    entity_type,
                    str(keyword),
                    normalized_keyword_tokens,
                ))

    # Longest phrases first prevents short keywords like "Java" from stealing
    # tokens inside longer entities like "Java Script" when both exist.
    keyword_specs.sort(key=lambda item: item[0], reverse=True)

    for keyword_len, entity_type, keyword, keyword_tokens in keyword_specs:
        b_tag = label2id[f"B-{entity_type}"]
        i_tag = label2id[f"I-{entity_type}"]

        for start in range(0, len(tokens) - keyword_len + 1):
            end = start + keyword_len
            if normalized_tokens[start:end] != keyword_tokens:
                continue

            existing = labels[start:end]
            if any(label != label2id["O"] for label in existing):
                existing_names = [id2label[label] for label in existing]
                target_names = [id2label[b_tag]] + [id2label[i_tag]] * (keyword_len - 1)
                if existing_names != target_names:
                    conflicts.append({
                        "keyword": keyword,
                        "entity_type": entity_type,
                        "span": (start, end),
                        "tokens": tokens[start:end],
                        "existing_labels": existing_names,
                        "target_labels": target_names,
                    })
                continue

            labels[start] = b_tag
            for pos in range(start + 1, end):
                labels[pos] = i_tag

    return {"tokens": tokens, "ner_tags": labels, "conflicts": conflicts}


def validate_bio_sequence(label_ids):
    previous_entity = None
    for label_id in label_ids:
        label = id2label[label_id]
        if label == "O":
            previous_entity = None
            continue
        prefix, entity = label.split("-", 1)
        if prefix == "B":
            previous_entity = entity
            continue
        if prefix == "I" and previous_entity == entity:
            continue
        return False
    return True


def summarize_bio_quality(df):
    conflict_count = sum(len(item.get("conflicts", [])) for item in df["parse_bio"])
    invalid_count = sum(not validate_bio_sequence(item["ner_tags"]) for item in df["parse_bio"])
    label_counter = Counter(
        id2label[label]
        for item in df["parse_bio"]
        for label in item["ner_tags"]
    )
    return {
        "rows": len(df),
        "conflicts": conflict_count,
        "invalid_bio_rows": invalid_count,
        "label_distribution": dict(label_counter),
    }
```

```python
def align_target(labels, word_ids):
    begin2inside = {
        1: 2,  # B-LANG (1) -> I-LANG (2)
        3: 4   # B-TECH (3) -> I-TECH (4)
    }

    align_labels = []
    last_word = None

    for word in word_ids:
        if word is None:
            label = -100
        elif word != last_word:
            label = labels[word]
        else:
            label = labels[word]
            if label in begin2inside:
                label = begin2inside[label]

        align_labels.append(label)
        last_word = word

    return align_labels
```

```python
# Extract labels and word_ids
labels = train_df.iloc[1]["parse_bio"]['ner_tags']
word_ids = token.word_ids()

# Use the align_target function to align labels
aligned_target = align_target(labels, word_ids)

print("labels:\n")
print(labels)

print("words id:\n")
print(word_ids)
# Print tokenized tokens, original labels, and aligned labels
print(token.tokens(), '\n--------------------------------------------------------------------------------------\n',
      labels, '\n--------------------------------------------------------------------------------------\n',
      aligned_target)
```

```python
def tokenize_fn_with_stride(batch):
    tokens_batch = [item["tokens"] for item in batch["parse_bio"]]
    ner_tags_batch = [item["ner_tags"] for item in batch["parse_bio"]]

    tokenized_inputs = tokenizer(
        tokens_batch,
        truncation=True,
        max_length=512,
        stride=50,
        return_overflowing_tokens=True,
        is_split_into_words=True,
    )

    aligned_targets_batch = []
    sample_mapping = tokenized_inputs.pop("overflow_to_sample_mapping")

    for chunk_index, original_sample_index in enumerate(sample_mapping):
        original_labels = ner_tags_batch[original_sample_index]
        word_ids = tokenized_inputs.word_ids(chunk_index)
        aligned_targets_batch.append(align_target(original_labels, word_ids))

    tokenized_inputs["labels"] = aligned_targets_batch
    return tokenized_inputs
```

```python
import numpy as np


def compute_metrics(logits_and_labels):
    logits, labels = logits_and_labels
    predictions = np.argmax(logits, axis=-1)

    str_labels = [
        [label_names[token_label] for token_label in label if token_label != -100]
        for label in labels
    ]
    str_preds = [
        [label_names[pred] for pred, token_label in zip(prediction, label) if token_label != -100]
        for prediction, label in zip(predictions, labels)
    ]

    results = metric.compute(predictions=str_preds, references=str_labels, zero_division=0)
    metrics = {
        "precision": results["overall_precision"],
        "recall": results["overall_recall"],
        "f1": results["overall_f1"],
        "accuracy": results["overall_accuracy"],
    }

    for entity_name in ["TECH_SKILL", "KNOW"]:
        if entity_name in results:
            metrics[f"{entity_name}_precision"] = results[entity_name]["precision"]
            metrics[f"{entity_name}_recall"] = results[entity_name]["recall"]
            metrics[f"{entity_name}_f1"] = results[entity_name]["f1"]
            metrics[f"{entity_name}_number"] = results[entity_name]["number"]

    return metrics
```

```python
def bio_entities(labels):
    entities = []
    start = None
    entity_type = None

    for index, label in enumerate(labels + ["O"]):
        if label == "O":
            if entity_type is not None:
                entities.append((entity_type, start, index))
                start = None
                entity_type = None
            continue

        prefix, current_type = label.split("-", 1)
        if prefix == "B" or current_type != entity_type:
            if entity_type is not None:
                entities.append((entity_type, start, index))
            start = index
            entity_type = current_type

    return entities


error_rows = []
for row_idx, (gold_labels, pred_labels) in enumerate(zip(eval_label_names, eval_pred_names)):
    if gold_labels == pred_labels:
        continue
    error_rows.append({
        "row_idx": row_idx,
        "gold_entities": bio_entities(gold_labels),
        "pred_entities": bio_entities(pred_labels),
        "gold_labels": gold_labels,
        "pred_labels": pred_labels,
    })

error_df = pd.DataFrame(error_rows)
error_path = "/content/drive/MyDrive/Classroom/DATN/Model/ner_eval_errors.csv"
error_df.to_csv(error_path, index=False)
print({"error_rows": len(error_df), "error_path": error_path})
error_df.head(20)
```
